```python
import pandas as pd
import geopandas as gpd
import folium
from ipywidgets import interact, widgets
from IPython.display import display
import branca.colormap as cm
from shapely import wkt
```

```python
#import data
df = pd.read_csv(r'C:\Users\ici-1\Desktop\NADJA\Verrio\visualize_growth_stability.csv')
```

```python
df.columns
```

```python
df['geometry'] = df['geometry'].apply(wkt.loads)
df = gpd.GeoDataFrame(df, geometry='geometry')
df.set_crs("EPSG:4326", inplace=True)
```

```python
print(type(df))
print(df.geometry.iloc[0])
```

```python
#only plot the bearing fields
df = df[df['status']=='bearing']
```

```python
@interact(year=widgets.IntSlider(
    min=df['year'].min(), max=df['year'].max(), step=1, value=df['year'].min()))
def show_map(year):
    filtered = df[df['year'] == year].copy()

    # Ensure GeoDataFrame with CRS
    if not isinstance(filtered, gpd.GeoDataFrame):
        filtered = gpd.GeoDataFrame(filtered, geometry='geometry')
    if filtered.crs is None:
        filtered.set_crs("EPSG:4326", inplace=True)

    # Calculate centroids for map centering
    centroids = filtered.geometry.centroid
    lat = centroids.y.max()
    lon = centroids.x.max()

    # Create color map
    # values = filtered['pct_diff_from_P95_avg']
    # colormap = cm.linear.YlOrRd_09.scale(values.min(), values.max())  # yellow→orange→red
    # colormap.caption = "Pct diff from P95 avg"


        # Define red → yellow → green ramp
    colormap = cm.LinearColormap(
        colors=["red", "yellow", "green"],
        vmin=df["pct_diff_from_P95_avg"].min(),
        vmax=df["pct_diff_from_P95_avg"].max()
    )
    colormap.caption = "Pct diff from P95 avg"

    # Create folium map
    m = folium.Map(location=[lat, lon], zoom_start=13, tiles="CartoDB positron")

    # Add GeoJson with color
    folium.GeoJson(
        filtered,
        name="fields",
        style_function=lambda feature: {
            'fillColor': colormap(feature['properties']['pct_diff_from_P95_avg']),
            'color': 'black',
            'weight': 1,
            'fillOpacity': 0.6
        },
        tooltip=folium.GeoJsonTooltip(fields=["pct_diff_from_P95_avg"])
    ).add_to(m)

    # Add color legend
    colormap.add_to(m)

    folium.LayerControl().add_to(m)
    display(m)
```

```python
def create_map_for_year(year, save_html=False):
    filtered = df[df['year'] == year].copy()

    # Ensure it's a valid GeoDataFrame with CRS
    if not isinstance(filtered, gpd.GeoDataFrame):
        filtered = gpd.GeoDataFrame(filtered, geometry='geometry')
    if filtered.crs is None:
        filtered.set_crs("EPSG:4326", inplace=True)

    # Calculate centroids
    centroids = filtered.geometry.centroid
    lat = centroids.y.mean()
    lon = centroids.x.mean()

    # Set up colormap
    values = filtered['pct_diff_from_P95_avg']
    colormap = cm.linear.YlOrRd_09.scale(values.min(), values.max())
    colormap.caption = "Pct diff from P95 avg"

    # Create folium map
    m = folium.Map(location=[lat, lon], zoom_start=13, tiles="CartoDB positron")

    folium.GeoJson(
        filtered,
        name="fields",
        style_function=lambda feature: {
            'fillColor': colormap(feature['properties']['pct_diff_from_P95_avg']),
            'color': 'black',
            'weight': 1,
            'fillOpacity': 0.6
        },
        tooltip=folium.GeoJsonTooltip(fields=["pct_diff_from_P95_avg"])
    ).add_to(m)

    colormap.add_to(m)
    folium.LayerControl().add_to(m)

    if save_html:
        m.save(f"map_{year}.html")
        print(f"Saved map to map_{year}.html")

    return m
```

```python
for yr in df['year'].unique():
    create_map_for_year(yr, save_html=True)
```

```python
# Your GeoDataFrame must have these columns
# - 'geometry' (valid geometries)
# - 'year'
# - 'pct_diff_from_P95_avg'

from folium import Element

# Get center of map from entire dataset
centroids = df.geometry.centroid
lat = centroids.y.mean()
lon = centroids.x.mean()

# Create base map
m = folium.Map(location=[lat, lon], zoom_start=13, tiles="CartoDB positron")

# Colormap setup
all_values = df['pct_diff_from_P95_avg']
colormap = cm.linear.YlOrRd_09.scale(all_values.min(), all_values.max())
colormap.caption = "Pct diff from P95 avg"
colormap.add_to(m)

# Dictionary to keep layers by year
year_layers = {}

for year in sorted(df['year'].unique()):
    filtered = df[df['year'] == year]

    layer = folium.FeatureGroup(name=str(year), show=False)

    folium.GeoJson(
        filtered,
        name=f"Fields {year}",
        style_function=lambda feature: {
            'fillColor': colormap(feature['properties']['pct_diff_from_P95_avg']),
            'color': 'black',
            'weight': 1,
            'fillOpacity': 0.6
        },
        tooltip=folium.GeoJsonTooltip(fields=["pct_diff_from_P95_avg"])
    ).add_to(layer)

    layer.add_to(m)
    year_layers[year] = layer

# Activate only the first year
first_year = sorted(year_layers.keys())[0]
year_layers[first_year].show = True

# Layer control
folium.LayerControl(collapsed=False).add_to(m)

# Create the dropdown HTML separately
dropdown_options = ''.join([f'<option value="{year}">{year}</option>' for year in sorted(year_layers.keys())])

dropdown_script = """
<select id="yearDropdown" onchange="switchYear(this.value)"
        style="position: fixed; top: 10px; left: 50px; z-index: 9999;">
  {options}
</select>

<script>
    function switchYear(year) {{
        const yearLayers = {years};
        map.eachLayer(function(layer) {{
            if (layer.options && layer.options.name && yearLayers.includes(layer.options.name)) {{
                if (layer.options.name === year) {{
                    map.addLayer(layer);
                }} else {{
                    map.removeLayer(layer);
                }}
            }}
        }});
    }}
</script>
""".format(
    options=dropdown_options,
    years=str([str(y) for y in year_layers.keys()])  # safely embed year list as JS array
)

# Add to map
m.get_root().html.add_child(Element(dropdown_script))
```

```python
# Save the map to HTML
m.save("interactive_map_with_dropdown.html")
print("Map saved to interactive_map_with_dropdown.html")
```
